## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The dataset contains 30,000 pages. Among them, 1,141 pages are in positions 1–3 having an average CTR of 2.71, while 7,273 pages are in positions 11–20 having an average CTR of 0.32. This difference suggests that search position and click-through performance are related in the dataset, making it useful to investigate which pages may deserve further review.

```python
!git clone https://github.com/adithyajupally/flyrank-ml-internship-jupally.git
%cd flyrank-ml-internship-jupally
```

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Rows:", len(df))
print("Columns:", len(df.columns))

print("Columns:", df.columns.tolist())

print("Average CTR:", round(df["ctr"].mean(), 2))
print("Average position:", round(df["avg_position"].mean(), 2))
print("Average engagement rate:", round(df["engagement_rate"].mean(), 2))

df["position_group"] = pd.cut(
    df["avg_position"],
    bins=[0, 3, 10, 20, 50, 100],
    labels=["1-3", "4-10", "11-20", "21-50", "51-100"]
)

print(df.groupby("position_group",observed=True)["ctr"].agg(["count", "mean"]))
```
