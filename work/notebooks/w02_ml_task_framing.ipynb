## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

My lane is a scoring task. The goal is to assign each page an opportunity score based on its search-performance signals, so pages that may be receiving fewer clicks than expected for their search position can be prioritized for review. I chose scoring because the output is a continuous priority signal rather than a simple yes/no class.

```python
task_type = "Scoring"
print("ML task type:", task_type)
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The target is observed CTR, which measures the proportion of impressions that result in clicks. For this lane, CTR is used as a measurable proxy for click performance, while the broader goal is to identify pages that may be receiving fewer clicks than expected for their search position. The opportunity score should use information available at decision time and avoid future or label-derived information that could cause leakage.

```python
!git clone https://github.com/adithyajupally/flyrank-ml-internship-jupally.git
%cd flyrank-ml-internship-jupally
```

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

target = "ctr"

print("Rows:", len(df))
print("Target:", target)
print("Target mean:", round(df[target].mean(), 3))
print("Target median:", round(df[target].median(), 3))
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

I will use Mean Absolute Error (MAE) as the success metric. Lower MAE means that the predicted CTR is closer to the observed CTR on average. This is a simple way to compare a model with a baseline.

```python
import numpy as np

y = df["ctr"].dropna()

baseline_prediction = y.mean()
baseline_mae = np.mean(np.abs(y - baseline_prediction))

print("Baseline MAE:", round(baseline_mae, 4))
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

The unit of analysis is one page/content item. Each row represents one page with its search-performance, engagement, and content-related signals. The opportunity score will therefore be assigned at the page level so that pages can be ranked for human review.

```python
page_df = df[["content_id","avg_position","ctr","impressions_90d","clicks_90d","engagement_rate","content_age_days","days_since_last_update"]].copy()

print("Rows:", len(page_df))
print("Columns:", page_df.columns.tolist())

page_df.head()
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule would be too simple because CTR can vary with several page and search-performance signals, not just one threshold. A rule such as flagging every page below a fixed CTR could miss pages that perform differently because of their search position, impressions, engagement, or content characteristics. ML can combine multiple signals to produce a more useful score for prioritizing pages for human review.

```python
signals = ["avg_position","impressions_90d","engagement_rate","content_age_days","days_since_last_update"]

print("Signals considered:")
for signal in signals:
    print("->", signal)
```
